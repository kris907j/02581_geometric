### Problem 1 (theoretical): Find the Optimal Translation ###
$\newcommand{\bbp}{\mathbf{p}}
\newcommand{\bbtp}{\tilde{\mathbf{p}}}
\newcommand{\bbt}{\mathbf{t}}
$
Let $\bbp_i=(x_i,y_i,z_i)^T$, $\bbtp_i=(\tilde{x}_i,\tilde{y}_i,\tilde{z}_i)^T$, for $i=1,\dots,n$ be corresponding points and let $\bbt=(a,b,c)^T$ be a translation vector of the $\bbp_i$ points. The Sum of Squared Distances between corresponding points is:
 
\begin{align*}
  SSD(a,b,c) &= \sum_{i=1}^{n}|| (\bbp_i +\bbt)-\bbtp_i)||^2 \\
   &= \sum_{i=1}^{n}\Bigl( (x_i +a-\tilde{x}_i)^2 + (y_i +b-\tilde{y}_i)^2 + (z_i +c-\tilde{z}_i)^2\Bigl).
\end{align*}
 
Note that $SSD$ is a non-negative second order polynomial in $a$, $b$ and $c$. Show that the only (and hence minimizing) stationary point for $SSD$ is when $$\bbt=1/n\bigl(\sum_{i=1}^n\bbtp_i-\sum_{i=1}^n \bbp_i\bigr).$$


We start by setting each of the partial derivatives of $SSD$ with respect to $a$, $b$, and $c$ to zero and solve for $a$, $b$, and $c$.

$\frac{\partial SSD}{\partial a} = \sum_{i=1}^n 2(x_i+a - \tilde{x}_i) = 0$ \
$\sum_{i=1}^n 2x_i +\sum_{i=1}^n 2a -\sum_{i=1}^n 2\tilde{x}_i = 0$ \
$\sum_{i=1}^n 2x_i +2an -\sum_{i=1}^n 2\tilde{x}_i = 0$\
$an = \sum_{i=1}^n \tilde{x}_i - \sum_{i=1}^n x_i$ \
$a=\frac{1}{n}\bigl(\sum_{i=1}^n \tilde{x}_i - \sum_{i=1}^n x_i)$

$\frac{\partial SSD}{\partial b} = \sum_{i=1}^n 2(y_i+b - \tilde{y}_i) = 0$ \
$b=\frac{1}{n}\bigl(\sum_{i=1}^n \tilde{y}_i - \sum_{i=1}^n y_i)$

$\frac{\partial SSD}{\partial c} = \sum_{i=1}^n 2(z_i+c - \tilde{z}_i) = 0$ \
$c=\frac{1}{n}\bigl(\sum_{i=1}^n \tilde{z}_i - \sum_{i=1}^n z_i)$

$t = (a,b,c) \
= \frac{1}{n}\bigl(\sum_{i=1}^n \tilde{x}_i - \sum_{i=1}^n x_i) + \frac{1}{n}\bigl(\sum_{i=1}^n \tilde{y}_i - \sum_{i=1}^n y_i) + \frac{1}{n}\bigl(\sum_{i=1}^n \tilde{z}_i - \sum_{i=1}^n z_i) \
= \frac{1}{n}\bigl(\sum_{i=1}^n\bbtp_i-\sum_{i=1}^n \bbp_i\bigr)$

```python
#   Problem 2: Compute Optimal Transformation
#
#     Insert below code which transforms MATCH_PTS to align as well as possible with ORIG_PTS
#     by computing the optimal translation and rotation.

MATCH_PTS = array([[1.8,  4.93], [1.17, 4.12], [2.78, 4.19], [4.28, 3.54], [6.45, 2.24],
                   [6.33, 3.22], [6.38, 4.6 ], [5.48, 4.12], [5.12, 7.57], [4.35, 7.84]])

#   ----- Enter code below ------
# t = 1/len(ORIG_PTS) * (mean(MATCH_PTS, axis=0) - mean(ORIG_PTS, axis=0))

# Move the points to the center of mass (making the optimal translation zero) (slide 22)
ORIG_PTS_centered = ORIG_PTS - mean(ORIG_PTS, axis=0) 
MATCH_PTS_centered = MATCH_PTS - mean(MATCH_PTS, axis=0)

# Compute the optimal rotation using SVD
S = ORIG_PTS_centered.T @ MATCH_PTS_centered # slide 26
U, Sigma, Vt = svd(S) # slide 26
R = U @ Vt # slide 27

# Apply the transformation to the matched points
transformed_pts = (R @ MATCH_PTS_centered.T).T + mean(ORIG_PTS, axis=0) #we start by rotating the points in origo and they add them to the original points to move from origo to the points.
#   ------ 

sfig = subplots.make_subplots(rows=1, cols=2, subplot_titles=('Before Alignment', 'After Alignment'))

# Subfigure 1: Plot the original points and matching points before alignment
sfig.add_scatter(x=ORIG_PTS[:,0],y=ORIG_PTS[:,1],mode='lines+markers', name='original points',col=1,row=1)
sfig.add_scatter(x=MATCH_PTS[:,0],y=MATCH_PTS[:,1],mode='lines+markers', name='match points',col=1,row=1)

# Subfigure 2: Plot the points and matches after alignment
sfig.add_scatter(x=ORIG_PTS[:,0],y=ORIG_PTS[:,1],mode='lines+markers', name='original points',col=2,row=1)
sfig.add_scatter(x=transformed_pts[:,0],y=transformed_pts[:,1],mode='lines+markers', name='aligned match points',col=2,row=1)


# Plot the lines connecting points to matches
corr = ndarray((ORIG_PTS.shape[0]*3,2))
for i in range(ORIG_PTS.shape[0]):
    corr[3*i:3*i+3] = array([ORIG_PTS[i], transformed_pts[i], [None, None]])
sfig.add_scatter(x=corr[:,0],y=corr[:,1],mode='lines',name='correspondences',col=2,row=1)

py.iplot(sfig)
```
